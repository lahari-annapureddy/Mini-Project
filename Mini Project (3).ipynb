```python
# Student Spending Behaviour Analysis using input and print statements

# Input
pocket_money = float(input("Enter your monthly pocket money: "))
food = float(input("Enter amount spent on food: "))
travel = float(input("Enter amount spent on travel: "))
entertainment = float(input("Enter amount spent on entertainment: "))
shopping = float(input("Enter amount spent on shopping: "))

# Calculations
total_spent = food + travel + entertainment + shopping
savings = pocket_money - total_spent

# Percentages
food_percent = (food / total_spent) * 100 if total_spent != 0 else 0
savings_percent = (savings / pocket_money) * 100 if pocket_money != 0 else 0

print("-----Analysis-----")
print("Total Spent:", total_spent)
print("Savings:", savings)
print("Food Spending %:", round(food_percent, 2), "%")
print("Savings %:", round(savings_percent, 2), "%")

# Suggestions
print("----- Suggestions -----")

if total_spent > pocket_money:
    print("You have exceeded your budget. Try reducing entertainment or shopping expenses.")
elif savings_percent < 20:
    print("Your savings are below 20%. Consider saving first before spending.")
elif food_percent > 40:
    print("You are spending a high portion on food. Try setting a weekly limit.")
else:
    print("Good financial management! Maintain this balanced spending pattern.")
```

```python
#Using statistics module to get mean and median using CSV file

import csv
import statistics

expenditures = []

# opening the csv file
with open("Data - Sheet1 (1).csv", "r") as file:
    reader = csv.DictReader(file)
    
    for row in reader:
        expenditures.append(float(row["Expenditure"]))


# calculating mean and median
Average_expenditure = statistics.mean(expenditures)
Median_expenditure = statistics.median(expenditures)

print("Expenditures:", expenditures)
print("Average Expenditure =", Average_expenditure)
print("Median Expenditure =", Median_expenditure)
```

```python
#Analysing expense of three random students from csv file
import pandas as pd

# Load CSV
file_path = "Data - Sheet1 (1).csv"
df = pd.read_csv(file_path)

# Clean numeric columns
columns_to_convert = ["Allowance", "Food", "Travelling", "Entertainment"]

for col in columns_to_convert:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(",", "")
        .str.replace("Nope|No|None", "0", regex=True)
    )
    df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)

# Randomly select 3 students
random_students = df.sample(n=3, random_state=None)

print("3 STUDENTS EXPENSE ANALYSIS ")

for index, row in random_students.iterrows():

    name = row["Names"]
    income = row["Allowance"]

    total_expense = row["Food"] + row["Travelling"] + row["Entertainment"]
    savings = income - total_expense

    print("Name:", {name})
    print("Allowance:", {income})
    print("Total Spending:", {total_expense})
    print("Remaining Balance", {savings})

    print("Suggestion:")

    if savings < 0:
        print("Overspending! Reduce expenses.")
    elif savings < income * 0.2:
        print("Try saving at least 20% of allowance.")
    elif row["Food"] > income * 0.5:
        print("High food expense. Budget meals.")
    elif row["Entertainment"] > income * 0.3:
        print("Entertainment spending is high.")
    else:
        print("Good financial management.")
```

```python
#Creating pie chart to showcase the overall categorial expense of all students
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV
df = pd.read_csv("Data - Sheet1 (1).csv")

# Convert columns to numeric
cols = ["Food", "Travelling", "Entertainment"]
for col in cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Calculate total spending per category
food_total = df["Food"].sum()
travel_total = df["Travelling"].sum()
ent_total = df["Entertainment"].sum()

# Labels and values
x = ["Food", "Travel", "Entertainment"]
y = [food_total, travel_total, ent_total]

# Pie chart
plt.pie(y, labels=x, autopct='%1.1f%%', rotatelabels=True)
plt.title("Overall Expenditure Distribution")
plt.show()
```

```python
#Graphical User Interaface for student expenditure analysis

import tkinter as tk
from tkinter import messagebox
import pandas as pd
import datetime


# Global Variables
expenses_dict = {}
MAX_ENTRIES = 25



# Function to Add Expense
def add_expense():
    if len(expenses_dict) >= MAX_ENTRIES:
        messagebox.showwarning("Limit Reached", "Maximum 25 entries allowed!")
        return

    category = category_entry.get()
    amount = amount_entry.get()

    if category == "" or amount == "":
        messagebox.showerror("Input Error", "Please fill all fields!")
        return

    try:
        amount = float(amount)
    except:
        messagebox.showerror("Input Error", "Amount must be in numeric values")
        return

    if category in expenses_dict:
        expenses_dict[category] += amount
    else:
        expenses_dict[category] = amount

    expense_listbox.insert(tk.END, f"{category} - ₹{amount}")

    category_entry.delete(0, tk.END)
    amount_entry.delete(0, tk.END)



# Function to Analyze Spending
def analyze_spending():
    if allowance_entry.get() == "":
        messagebox.showerror("Error", "Enter Monthly Allowance!")
        return

    try:
        allowance = float(allowance_entry.get())
    except:
        messagebox.showerror("Error", "Allowance must be numeric!")
        return

    total_spent = sum(expenses_dict.values())
    savings = allowance - total_spent
    savings_percent = (savings / allowance) * 100

    result_text.delete("1.0", tk.END)

    result_text.insert(tk.END, "------ Spending Analysis ------\n\n")
    result_text.insert(tk.END, f"Total Allowance: ₹{allowance}\n")
    result_text.insert(tk.END, f"Total Spent: ₹{total_spent}\n")
    result_text.insert(tk.END, f"Savings: ₹{savings}\n")
    result_text.insert(tk.END, f"Savings Percentage: {round(savings_percent,2)}%\n\n")

    result_text.insert(tk.END, "Category Breakdown:\n")

    for category, amount in expenses_dict.items():
        percent = (amount / allowance) * 100
        result_text.insert(tk.END, f"{category}: ₹{amount} ({round(percent,2)}%)\n")

    result_text.insert(tk.END, "\n------ Suggestions ------\n")

    if total_spent > 0.95 * allowance:
        result_text.insert(tk.END, "You are overspending! Reduce non-essential expenses.\n")

    if savings_percent < 10:
        result_text.insert(tk.END, "Try to save at least 10% of your allowance.\n")

    if savings_percent >= 10 and savings_percent < 25:
        result_text.insert(tk.END, "Good job! You are maintaining decent savings.\n")

    if savings_percent >= 25:
        result_text.insert(tk.END, "Excellent financial discipline!\n")

    if "Entertainment" in expenses_dict and expenses_dict["Entertainment"] > 0.3 * allowance:
        result_text.insert(tk.END, "Consider reducing entertainment expenses.\n")

    # Create Pandas DataFrame
    df = pd.DataFrame(list(expenses_dict.items()), columns=["Category", "Amount"])

    # Save CSV
    df.to_csv("student_spending_data.csv", index=False)

    # Save Report to Text File
    with open("student_spending_report.txt", "w") as file:
        file.write("Student Spending Behaviour Report\n")
        file.write("Generated on: " + str(datetime.datetime.now()) + "\n\n")
        file.write(f"Allowance: ₹{allowance}\n")
        file.write(f"Total Spent: ₹{total_spent}\n")
        file.write(f"Savings: ₹{savings}\n")
        file.write(f"Savings Percentage: {round(savings_percent,2)}%\n\n")
        file.write("Category Details:\n")
        file.write(df.to_string())

    messagebox.showinfo("Success", "Analysis Completed & Report Saved!")



# Function to Clear Data
def clear_data():
    expenses_dict.clear()
    expense_listbox.delete(0, tk.END)
    result_text.delete("1.0", tk.END)
    messagebox.showinfo("Reset", "All data cleared!")



# GUI Window Setup
root = tk.Tk()
root.title("Student Spending Behaviour Analysis System")
root.geometry("650x750")
root.configure(bg="#f0f4f7")

# Title
tk.Label(root, text="Student Spending Behaviour Analysis",
         font=("Arial", 16, "bold"),
         bg="#f0f4f7").pack(pady=10)

# Student Name
tk.Label(root, text="Student Name:", bg="#f0f4f7").pack()
name_entry = tk.Entry(root, width=40)
name_entry.pack()

# Allowance
tk.Label(root, text="Monthly Allowance:", bg="#f0f4f7").pack()
allowance_entry = tk.Entry(root, width=40)
allowance_entry.pack()

# Category Input
tk.Label(root, text="Expense Category:", bg="#f0f4f7").pack()
category_entry = tk.Entry(root, width=40)
category_entry.pack()

# Amount Input
tk.Label(root, text="Amount:", bg="#f0f4f7").pack()
amount_entry = tk.Entry(root, width=40)
amount_entry.pack()

# Buttons
tk.Button(root, text="Add Expense", width=20, bg="#4CAF50", fg="black",
          command=add_expense).pack(pady=5)

tk.Button(root, text="Analyze Spending", width=20, bg="#2196F3", fg="black",
          command=analyze_spending).pack(pady=5)

tk.Button(root, text="Clear Data", width=20, bg="#f44336", fg="black",
          command=clear_data).pack(pady=5)

# Expense List
tk.Label(root, text="Expense Entries:", bg="#f0f4f7").pack()
expense_listbox = tk.Listbox(root, width=60, height=10)
expense_listbox.pack(pady=5)

# Result Section
tk.Label(root, text="Analysis Report:", bg="#f0f4f7").pack()
result_text = tk.Text(root, height=15, width=75)
result_text.pack(pady=5)

root.mainloop()
```
